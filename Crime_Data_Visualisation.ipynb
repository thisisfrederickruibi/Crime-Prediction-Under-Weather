# **Crime Data Visualisation**

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the CSV files
robbery_df = pd.read_csv('Robbery.csv')
burglary_df = pd.read_csv('Burglary.csv')
theft_df = pd.read_csv('Theft.csv')
```

```python
# Convert 'Date' column to datetime
robbery_df['Date'] = pd.to_datetime(robbery_df['Date'])
burglary_df['Date'] = pd.to_datetime(burglary_df['Date'])
theft_df['Date'] = pd.to_datetime(theft_df['Date'])
```

```python
# List of boroughs 
boroughs = ['Camden', 'City of London', 'Greenwich', 'Hackney', 'Hammersmith and Fulham', 
            'Islington', 'Kensington and Chelsea', 'Lambeth', 'Lewisham', 'Southwark', 
            'Tower Hamlets', 'Wandsworth', 'Westminster']
```

```python
# Define date ranges
period1_start = '2019-01-01'
period1_end = '2019-06-30'
period2_start = '2019-07-01'
period2_end = '2019-12-31'
period3_start = '2020-01-01'
period3_end = '2020-06-30'
```

```python
# Filter data for the three periods
robbery_period1 = robbery_df[(robbery_df['Date'] >= period1_start) & (robbery_df['Date'] <= period1_end)]
burglary_period1 = burglary_df[(burglary_df['Date'] >= period1_start) & (burglary_df['Date'] <= period1_end)]
theft_period1 = theft_df[(theft_df['Date'] >= period1_start) & (theft_df['Date'] <= period1_end)]

robbery_period2 = robbery_df[(robbery_df['Date'] >= period2_start) & (robbery_df['Date'] <= period2_end)]
burglary_period2 = burglary_df[(burglary_df['Date'] >= period2_start) & (burglary_df['Date'] <= period2_end)]
theft_period2 = theft_df[(theft_df['Date'] >= period2_start) & (theft_df['Date'] <= period2_end)]

robbery_period3 = robbery_df[(robbery_df['Date'] >= period3_start) & (robbery_df['Date'] <= period3_end)]
burglary_period3 = burglary_df[(burglary_df['Date'] >= period3_start) & (burglary_df['Date'] <= period3_end)]
theft_period3 = theft_df[(theft_df['Date'] >= period3_start) & (theft_df['Date'] <= period3_end)]
```

```python
# Function to plot crime data with bold and centered title
def plot_crime_data(df, crime_type, ax, period_label):
    for borough in boroughs:
        ax.plot(df['Date'], df[borough], label=borough)
    ax.set_title(f'{crime_type} Crimes in London Boroughs ({period_label})', 
                 fontweight='bold', loc='center', fontsize=12)
    ax.set_xlabel('Date')
    ax.set_ylabel('Number of Crimes')
    ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    ax.grid(True)
    ax.tick_params(axis='x', rotation=45)
```

```python
# Create figure for Period 1 (Jan 1, 2019 - Jun 30, 2019)
fig1, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(12, 15), sharex=True)
plot_crime_data(robbery_period1, 'Robbery', ax1, 'Jan 2019 - Jun 2019')
plot_crime_data(burglary_period1, 'Burglary', ax2, 'Jan 2019 - Jun 2019')
plot_crime_data(theft_period1, 'Theft', ax3, 'Jan 2019 - Jun 2019')
plt.tight_layout()
fig1.suptitle('Crime Data Visualisation (Jan 2019 - Jun 2019)', fontweight='bold', fontsize=14, y=1.02)
fig1.savefig('crime_plots_period1.png', bbox_inches='tight', dpi=300)
```

```python
# Create figure for Period 2 (Jul 1, 2019 - Dec 31, 2019)
fig2, (ax4, ax5, ax6) = plt.subplots(3, 1, figsize=(12, 15), sharex=True)
plot_crime_data(robbery_period2, 'Robbery', ax4, 'Jul 2019 - Dec 2019')
plot_crime_data(burglary_period2, 'Burglary', ax5, 'Jul 2019 - Dec 2019')
plot_crime_data(theft_period2, 'Theft', ax6, 'Jul 2019 - Dec 2019')
plt.tight_layout()
fig2.suptitle('Crime Data Visualisation (Jul 2019 - Dec 2019)', fontweight='bold', fontsize=14, y=1.02)
fig2.savefig('crime_plots_period2.png', bbox_inches='tight', dpi=300)
```

```python
# Create figure for Period 3 (Jan 1, 2020 - Jun 30, 2020)
fig3, (ax7, ax8, ax9) = plt.subplots(3, 1, figsize=(12, 15), sharex=True)
plot_crime_data(robbery_period3, 'Robbery', ax7, 'Jan 2020 - Jun 2020')
plot_crime_data(burglary_period3, 'Burglary', ax8, 'Jan 2020 - Jun 2020')
plot_crime_data(theft_period3, 'Theft', ax9, 'Jan 2020 - Jun 2020')
plt.tight_layout()
fig3.suptitle('Crime Data Visualisation (Jan 2020 - Jun 2020)', fontweight='bold', fontsize=14, y=1.02)
fig3.savefig('crime_plots_period3.png', bbox_inches='tight', dpi=300)
```
